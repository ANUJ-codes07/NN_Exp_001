To familiarize students with popular deep learning libraries and perform basic tensor creation and mathematical operations using TensorFlow.

```python
import tensorflow as tf
import numpy as np

print("TensorFlow version:", tf.__version__)
print("NumPy version:", np.__version__)
```

```python
import tensorflow as tf

print("TensorFlow is successfully installed.")
print("TensorFlow version:", tf.__version__)
```

```python
scalar = tf.constant(10)

print("Scalar Tensor:")
print(scalar)
```

```python
matrix = tf.constant([
    [1, 2, 3],
    [4, 5, 6]
])

print("Matrix Tensor:")
print(matrix)
```

```python
a = tf.constant([1, 2, 3])
b = tf.constant([4, 5, 6])

addition = tf.add(a, b)

print("Tensor Addition:")
print(addition)
```

```python
subtraction = tf.subtract(b, a)

print("Tensor Subtraction:")
print(subtraction)
```

```python
multiplication = tf.multiply(a, b)

print("Tensor Multiplication:")
print(multiplication)
```

```python
matrix_a = tf.constant([
    [1, 2],
    [3, 4]
])

matrix_b = tf.constant([
    [5, 6],
    [7, 8]
])

matrix_multiplication = tf.matmul(matrix_a, matrix_b)

print("Matrix Multiplication:")
print(matrix_multiplication)
```

```python
tensor = tf.constant([1, 2, 3, 4, 5, 6])

reshaped_tensor = tf.reshape(tensor, (2, 3))

print("Original Tensor:")
print(tensor)

print("\nReshaped Tensor:")
print(reshaped_tensor)
```

```python
numpy_array = np.array([10, 20, 30, 40, 50])

tensorflow_tensor = tf.convert_to_tensor(numpy_array)

print("NumPy Array:")
print(numpy_array)

print("\nTensorFlow Tensor:")
print(tensorflow_tensor)
```

## Comparison of Deep Learning Frameworks

| Feature | TensorFlow | PyTorch | Keras |
|---|---|---|---|
| Developer | Google | Meta | Originally François Chollet / now part of TensorFlow ecosystem |
| Type | Deep Learning Framework | Deep Learning Framework | High-level Deep Learning API |
| Ease of Use | Moderate | Easy to Moderate | Very Easy |
| Computation | Static and Dynamic Graph Support | Dynamic Computation Graph | High-level abstraction |
| Deployment | Excellent | Excellent | Excellent through TensorFlow ecosystem |
| Main Applications | Deep Learning, Production, Computer Vision, NLP | Research, Deep Learning, Computer Vision, NLP | Rapid Model Development and Prototyping |
| GPU Support | Yes | Yes | Yes |
| Community | Very Large | Very Large | Very Large |
| Major Advantage | Production and deployment | Flexibility and research | Simplicity and rapid development |

### TensorFlow
- Developed by Google.
- Suitable for large-scale deep learning applications.
- Provides strong deployment and production capabilities.
- Supports GPU and TPU acceleration.
- Used in computer vision, NLP, recommendation systems, and other AI applications.

### PyTorch
- Developed by Meta.
- Popular in research and experimentation.
- Provides flexible and intuitive tensor operations.
- Supports GPU acceleration.
- Widely used in computer vision, NLP, and research projects.

### Keras
- High-level deep learning API.
- Provides a simple and user-friendly interface.
- Makes neural network development faster.
- Commonly used for education, prototyping, and rapid model development.
- Integrated closely with TensorFlow.
